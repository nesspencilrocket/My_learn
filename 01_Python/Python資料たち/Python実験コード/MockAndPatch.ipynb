```python
import datetime
from unittest.mock import patch, MagicMock

# 1. テスト対象の関数（現在時刻を使用するもの）
# 通常は別ファイルにありますが、ここではデモ用に同じ場所に定義します
def print_current_time():
    # ここで datetime.datetime.now() が呼ばれます
    now = datetime.datetime.now()
    print(f"出力結果: {now}")
    return now

# --- 以下、モックとパッチを使ったテストコード ---

def run_test():
    # 2. 固定したい日時を作成
    target_dt = datetime.datetime(2025, 8, 31, 11, 45, 14)

    # 3. patchを使って datetime.datetime をモックに置き換える
    # 'datetime.datetime' は組み込み型で直接書き換えられないため、
    # ターゲットのモジュール内でインポートされている名前空間をパッチします。
    # このスクリプト内(__main__)にある datetime クラスを置き換えます。

    with patch('__main__.datetime.datetime') as mock_datetime:
        # mock_datetime.now() が呼ばれたら target_dt を返すように設定
        mock_datetime.now.return_value = target_dt

        # 副作用: 比較演算などが動くように継承元の性質を持たせる場合
        # (単純な出力だけなら上の return_value だけで十分です)
        mock_datetime.side_effect = lambda *args, **kwargs: datetime.datetime(*args, **kwargs)

        # 実行
        print("--- テスト実行開始 ---")
        print_current_time()
        print("--- テスト実行終了 ---")

if __name__ == '__main__':
    run_test()
```

解説とポイント
target_dt の作成

出力させたい「2025年8月31日 11:45:14」を本物の datetime オブジェクトとして作成しておきます。

patch の対象

datetime モジュールは C言語で書かれた拡張モジュールのため、datetime.datetime クラス自体を直接書き換えることはできません。

そのため、「その datetime を使っている場所（名前空間）」 をパッチします。上記の例では __main__ 内で使用している datetime.datetime を置き換えています。

return_value の設定

mock_datetime.now.return_value = target_dt とすることで、.now() メソッドが呼ばれた瞬間に、用意しておいた固定の日時オブジェクトを返させます。
