```python
from datetime import datetime, timedelta

now = datetime.now()

# 1. 未来へ：「1週間後」を計算
# timedelta(days=7) という「7日間の塊」を足す
next_week = now + timedelta(days=7)
print(f"来週: {next_week}")

# 2. 過去へ：「3時間前」を計算
# timedelta(hours=3) を引く
past_time = now - timedelta(hours=3)
print(f"3時間前: {past_time}")

# 3. 差分：日付同士を引き算すると「期間」が出る
# 画像では2023年ですが、動作確認用に2025年に修正しました
christmas = datetime(2025, 12, 25)
diff = christmas - now
print(f"クリスマスまであと {diff.days} 日！")
```

```python
from datetime import datetime

now = datetime.now()

# %Y=4桁の年, %m=月, %d=日, %H=時, %M=分
date_str = now.strftime("%Y年%m月%d日 %H:%M")

print(date_str)
# 出力例: 2025年12月10日 13:45
```

```python
from datetime import datetime, timedelta, timezone

# 1. 日本時間（JST）のルールを作る：UTCに9時間を足す
JST = timezone(timedelta(hours=+9), 'JST')

# 2. 現在時刻を取得するときに、JSTを指定する
now = datetime.now(JST)

# --- 以下は元のコードと同じ ---

# 文字列に変換
date_str = now.strftime("%Y年%m月%d日 %H:%M")
print(f"日本時間: {date_str}")

# 未来の計算（来週）
next_week = now + timedelta(days=7)
print(f"来週: {next_week.strftime('%Y/%m/%d %H:%M')}")
```
